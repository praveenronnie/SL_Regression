```python
import pandas as pd

dataset = pd.read_csv('insurance_pre.csv')
dataset = pd.get_dummies(dataset, dtype= int)
dataset
```

```python
ind_var = dataset.drop("charges", axis=1)
dep_var = dataset["charges"]
ind_var
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(ind_var, dep_var, test_size = 0.3, random_state= 42)
```

```python
from sklearn.preprocessing import StandardScaler

# Standard Scaling for SVM
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
def modelScore(model, x_train, x_test, y_train, y_test):

    from sklearn.metrics import r2_score

    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    r_score = r2_score(y_test , y_pred)
    return r_score
```

```python
def decisionTreeModel(x_train, x_test, y_train, y_test):
    from sklearn.tree import DecisionTreeRegressor

    dtRegressor = DecisionTreeRegressor(
        criterion="absolute_error", 
        splitter = "best", 
        max_depth= 5, 
        random_state=42
        )
    
    model_score = modelScore(dtRegressor, x_train, x_test, y_train, y_test)
    return model_score
```

```python
def randomForestModel(x_train, x_test, y_train, y_test):

    from sklearn.ensemble import RandomForestRegressor
    random_forest_model = RandomForestRegressor(
        criterion = 'squared_error',
        n_estimators= 100,
        max_depth = 10,
        max_features = 'sqrt',
        random_state = 0
    )
    rf_model_score = modelScore(random_forest_model, x_train, x_test, y_train, y_test)
    return rf_model_score
```

```python
def linearModel(x_train, x_test, y_train, y_test):
    from sklearn.linear_model import LinearRegression

    linear_model = LinearRegression()
    linear_score = modelScore(linear_model, x_train, x_test, y_train, y_test)
    return linear_score
```

```python
def svmModel(x_train, x_test, y_train, y_test):
    from sklearn.svm import SVR

    svr_model = SVR(kernel= "linear", C=1000, gamma= 3)
    svr_score = modelScore(svr_model, X_train, X_test, y_train, y_test)
    return svr_score
```

```python
linear_score = linearModel(X_train, X_test, y_train, y_test)
svm_score = svmModel(X_train, X_test, y_train, y_test)
decision_tree_score = decisionTreeModel(X_train, X_test, y_train, y_test)
random_forest_score = randomForestModel(X_train, X_test, y_train, y_test)

print(linear_score, '\n', svm_score, '\n', decision_tree_score, '\n', random_forest_score)
```

```python
def gradientBoostingModel(X_train, X_test, y_train, y_test):

    from sklearn.ensemble import GradientBoostingRegressor

    booster = GradientBoostingRegressor()
    gradient_score = modelScore(booster, X_train, X_test, y_train, y_test)
    return gradient_score

g_score = gradientBoostingModel(X_train, X_test, y_train, y_test)
print(g_score)
```

```python
def adaBoostingModel(X_train, X_test, y_train, y_test):

    from sklearn.ensemble import AdaBoostRegressor

    ada_boost = AdaBoostRegressor(
        n_estimators=200,
        learning_rate=0.1,
        random_state=42
    )
    ada_score = modelScore(ada_boost, X_train, X_test, y_train, y_test)
    return ada_score

g_score = adaBoostingModel(X_train, X_test, y_train, y_test)
print(g_score)
```

```python
def xgBoostingModel(X_train, X_test, y_train, y_test):

    from xgboost import XGBRegressor

    xg_boost = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=4,
    random_state=42
)

    xg_score = modelScore(xg_boost, X_train, X_test, y_train, y_test)
    return xg_score

g_score = xgBoostingModel(X_train, X_test, y_train, y_test)
print(g_score)
```
